```python
import numpy as np
arr = np.array([1,0,0,8])
print(arr)
np.save('my_array.npy', arr) # this is to save the data in array
```

```python
loaded_arr = np.load('my_array.npy') # this is to load the data in array
print(loaded_arr)
```
